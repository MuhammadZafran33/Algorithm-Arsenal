## 9. Handling Bad / Malformed Lines — `on_bad_lines`

Real-world CSVs sometimes have rows with the wrong number of fields, which normally raises an error. `on_bad_lines='skip'` tells pandas to silently drop those rows instead of failing (`on_bad_lines='warn'` does the same but prints a warning for each one).

```python
import pandas as pd

# 'skip' ignores the bad lines entirely. 
# You can also use 'warn' to skip them but print a warning for each one.
df = pd.read_csv('/kaggle/input/datasets/muhammadzafranai/diabetes-data/diabetes_dataset.csv', on_bad_lines='skip')

# Save the cleaned data to a new CSV
df.to_csv('cleaned_file.csv', index=False)
```

## 10. Setting Column Data Types — `dtype`

## 12. Converters — Transforming Values While Reading

`converters` applies a custom function to every value in a column **as the file is read**, before it even becomes part of the DataFrame. Here, a `rename()` function is used to shorten a long team name to its abbreviation.

```python
def rename(name):
    if name == "Royal Challengers Bangalore":
        return "RCB"
    else:
        return
```

```python
def rename(name):
    if name == "Royal Challengers Bangalore":
        return "RCB"
    else:
        return name  # <-- This tells Pandas to keep the original name
```

```python
rename("Royal Challengers Bangalore")
```

### CONVERTORS


```python
df4 = pd.read_csv('/kaggle/input/datasets/patrickb1912/ipl-complete-dataset-20082020/matches.csv', converters={'team1':rename})
df4
```

The same idea, applied to **multiple columns at once** by passing a dictionary that maps each column name to the same converter function:

```python
df4 = pd.read_csv(
    '/kaggle/input/datasets/patrickb1912/ipl-complete-dataset-20082020/matches.csv', 
    converters={
        'team1': rename, 
        'team2': rename,
        'toss_winner': rename,
        'winner': rename
    }
)
df4
```

An alternative to `converters`: read the file normally first, then loop over the target columns and apply the function with `.apply()` afterward. This is often simpler to read, at the cost of a small amount of extra memory/time versus converting while reading.

```python
# 1. Load the CSV normally
df4 = pd.read_csv('/kaggle/input/datasets/patrickb1912/ipl-complete-dataset-20082020/matches.csv')

# 2. List the columns you want to change
team_columns = ['team1', 'team2', 'toss_winner', 'winner']

# 3. Apply your rename function to all of them using a quick loop
for col in team_columns:
    df4[col] = df4[col].map(rename)
df4
```

## 13. Reading Large Files in Chunks — `chunksize`

For files too large to comfortably fit in memory, `chunksize` makes `read_csv()` return an **iterator** of DataFrames instead of one big DataFrame — each chunk has at most `chunksize` rows. You then process each chunk in a loop.

```python
dfs = pd.read_csv('/kaggle/input/datasets/patrickb1912/ipl-complete-dataset-20082020/matches.csv', chunksize = 400)
```

```python
for chunks in dfs:
    print(chunks.shape)
```

## 14. Alternative: Manual Iteration with `iterator=True`

`iterator=True` is similar to `chunksize`, but instead of automatically looping through fixed-size chunks, it gives you a reader object you can pull chunks from **on demand** with `.get_chunk(n)` — useful when you don't want to process the whole file, or want variable chunk sizes.

```python
reader = pd.read_csv(
    '/kaggle/input/datasets/patrickb1912/ipl-complete-dataset-20082020/matches.csv',
    iterator=True
)

first_100 = reader.get_chunk(100)   # pull the first 100 rows
print(first_100.shape)

next_50 = reader.get_chunk(50)      # pull the NEXT 50 rows, continuing from where we left off
print(next_50.shape)
```

## 15. Saving a DataFrame to CSV — `to_csv()`

Once you've cleaned or transformed a DataFrame, `to_csv()` writes it back out to disk. A few parameters worth knowing:
- `index=False` — don't write the DataFrame's row index as a column (usually what you want, otherwise you get an extra unnamed column when you re-read the file).
- `columns=[...]` — only write a subset of columns.
- `sep=` — use a delimiter other than a comma.

```python
# Default: writes every column, including the index, as a new column
df1.to_csv('diabetes_export.csv')

# Recommended for most cases: drop the index column
df1.to_csv('diabetes_export_no_index.csv', index=False)

# Write only specific columns
df1.to_csv('diabetes_subset.csv', columns=['Age', 'BMI', 'Outcome'], index=False)

# Confirm it round-trips correctly
pd.read_csv('diabetes_export_no_index.csv').head()
```

## 16. Custom Delimiters — `sep` / `delimiter`

Not every "CSV" actually uses a comma. Files exported from some tools use semicolons, tabs, or pipes instead. `sep` (an alias `delimiter` also exists) tells pandas which character separates the fields.

*(The example below builds a tiny CSV in memory with `io.StringIO` so it runs without needing an external file — the same `sep=` argument works identically with a real file path.)*

```python
from io import StringIO

raw_data = "name;age;city\nAli;22;Peshawar\nSara;25;Lahore\nUsman;28;Bannu"

df_semicolon = pd.read_csv(StringIO(raw_data), sep=';')
df_semicolon
```

## 17. Handling Missing Values — `na_values`

Real-world CSVs mark missing data in inconsistent ways: `?`, `NA`, `-`, empty strings, `"N/A"`, etc. Pandas already recognizes some common ones automatically, but `na_values` lets you tell it about custom placeholders in your specific dataset so they get converted to proper `NaN` values instead of being read as text.

```python
raw_data = "name,age,city\nAli,22,Peshawar\nSara,?,Lahore\nUsman,NA,-"

df_missing = pd.read_csv(StringIO(raw_data), na_values=['?', 'NA', '-'])
df_missing
```

```python
# Count how many missing values ended up in each column
df_missing.isnull().sum()
```

## 18. File Encoding — `encoding`

Pandas assumes UTF-8 by default. Files exported from Excel on Windows, or older systems, are sometimes saved in a different encoding (e.g. `'latin1'` / `'cp1252'`) — reading them without specifying `encoding` can raise a `UnicodeDecodeError` or silently produce garbled characters for accented letters.

```python
from io import BytesIO

raw_bytes = "city,notes\ncafé,münchen\n".encode('latin1')

df_encoded = pd.read_csv(BytesIO(raw_bytes), encoding='latin1')
df_encoded
```

## 19. Skipping Comment Lines — `comment`

Some CSVs (especially ones generated by scientific tools) include comment/metadata lines starting with a character like `#`. `comment='#'` tells pandas to ignore any line starting with that character, rather than trying to parse it as data.

```python
raw_data = (
    "# exported on 2026-08-31\n"
    "name,age\n"
    "Ali,22\n"
    "# next row is a duplicate for testing\n"
    "Sara,25\n"
)

df_comment = pd.read_csv(StringIO(raw_data), comment='#')
df_comment
```

## 20. Reading Compressed CSV Files — `compression`

Pandas can read `.csv.gz`, `.zip`, `.bz2`, and `.xz` files directly, without manually decompressing them first. `compression='infer'` (the default) figures it out from the file extension; you can also set it explicitly.

```python
import gzip

raw_csv_bytes = "name,age\nAli,22\nSara,25".encode('utf-8')
compressed = gzip.compress(raw_csv_bytes)

df_gz = pd.read_csv(BytesIO(compressed), compression='gzip')
df_gz

# With a real file on disk this is even simpler — pandas infers the compression from the extension:
# df = pd.read_csv('some_file.csv.gz')
```
