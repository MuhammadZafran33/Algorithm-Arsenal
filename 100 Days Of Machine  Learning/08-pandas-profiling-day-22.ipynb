### Viewing the Report Inline, Without Leaving the Notebook

`to_file()` (above) writes a standalone HTML file you'd open in a browser separately. `to_notebook_iframe()` embeds the same interactive report directly inside the notebook itself — often more convenient while actively exploring a new dataset.

```python
prof.to_notebook_iframe()
```

### Speeding It Up on Larger Datasets — `minimal=True`

By default, the report computes every possible statistic and correlation, including some expensive pairwise ones — this can get slow on datasets with many rows or many columns. `minimal=True` skips the most expensive computations (like certain correlation types) for a much faster report, at the cost of a bit of detail.

```python
prof_fast = ProfileReport(df, minimal=True)
prof_fast.to_file(output_file='output_minimal.html')
```

## Summary

| Task | Method |
|---|---|
| Generate a full report | `ProfileReport(df)` |
| Save it as a standalone HTML file | `prof.to_file('output.html')` |
| View it inline in the notebook | `prof.to_notebook_iframe()` |
| Speed it up on a large dataset | `ProfileReport(df, minimal=True)` |
