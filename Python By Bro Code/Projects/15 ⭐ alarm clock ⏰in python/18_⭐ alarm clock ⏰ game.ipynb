```python
import datetime
import time
import pygame

def set_alarm(alarm_time):
    # Ensure standard HH:MM:SS format with leading zeros if user typed e.g. 7:5:0
    try:
        alarm_time = datetime.datetime.strptime(alarm_time, "%H:%M:%S").strftime("%H:%M:%S")
    except ValueError:
        print("Invalid time format! Please use HH:MM:SS (e.g., 07:30:00 or 14:05:00).")
        return

    print(f"Alarm set for {alarm_time}")
    sound_file = "C:/Users/pts/OneDrive/Desktop/ML Proj/Courses/Bro Code Python/03-Object-Oriented-Programming/mat-jana-tum.mp3"

    is_running = True

    while is_running:
        current_time = datetime.datetime.now().strftime("%H:%M:%S")
        print(f"\rCurrent Time: {current_time}", end="")

        if current_time == alarm_time:
            print("\nWAKE UP! 😫")

            pygame.mixer.init()
            pygame.mixer.music.load(sound_file)
            pygame.mixer.music.play()

            # Press Enter to stop the audio immediately
            input("Press Enter to stop the alarm...")
            pygame.mixer.music.stop()

            is_running = False

        time.sleep(1)

if __name__ == "__main__":
    alarm_input = input("Enter the alarm time (HH:MM:SS): ")
    set_alarm(alarm_input)
```
