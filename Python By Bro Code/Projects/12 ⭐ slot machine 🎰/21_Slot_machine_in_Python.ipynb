```python
import random
import time

def spin_row():
    """Returns a list of 3 random symbols."""
    symbols = ['🍒', '🍉', '🍋', '🔔', '⭐']
    return [random.choice(symbols) for _ in range(3)]

def print_row(row):
    """Prints the row in a clean slot machine format."""
    print("*************")
    print(f" {row[0]} | {row[1]} | {row[2]} ")
    print("*************")

def get_payout(row, bet):
    """Checks for matching symbols and returns the payout multiplier."""
    if row[0] == row[1] == row[2]:
        if row[0] == '🍒':
            return bet * 3
        elif row[0] == '🍉':
            return bet * 4
        elif row[0] == '🍋':
            return bet * 5
        elif row[0] == '🔔':
            return bet * 10
        elif row[0] == '⭐':
            return bet * 20
    return 0

def main():
    balance = 100

    print("************************")
    print(" Welcome to Python Slots ")
    print("Symbols: 🍒 🍉 🍋 🔔 ⭐")
    print("************************")

    while balance > 0:
        print(f"\nCurrent balance: ${balance}")
        bet_input = input("Place your bet amount (or type 'q' to quit): ")

        if bet_input.lower() == 'q':
            break
        
        if not bet_input.isdigit():
            print("Please enter a valid whole number.")
            continue
            
        bet = int(bet_input)

        if bet > balance:
            print("Insufficient funds! You can't bet more than you have.")
            continue
        elif bet <= 0:
            print("Bet must be greater than $0.")
            continue

        # Subtract bet from balance
        balance -= bet

        # Spin the slots
        print("Spinning...")
        time.sleep(1) # Adds a small delay for suspense
        row = spin_row()
        print_row(row)

        # Calculate winnings
        payout = get_payout(row, bet)

        if payout > 0:
            print(f"JACKPOT! You won ${payout}!")
            balance += payout
        else:
            print("Sorry, you lost this round.")

    print("************************")
    print(f" Game over! You walked away with ${balance} ")
    print("************************")

if __name__ == '__main__':
    main()
```
