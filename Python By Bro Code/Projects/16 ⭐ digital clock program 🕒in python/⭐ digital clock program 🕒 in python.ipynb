```python
import sys
from PyQt5.QtWidgets import QApplication, QWidget, QLabel, QVBoxLayout
from PyQt5.QtCore import QTimer, QTime, Qt
from PyQt5.QtGui import QFont, QFontDatabase

class DigitalClock(QWidget):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("Digital Clock")
        self.initUI()

    def initUI(self):
        # 1. Set up Layout and Label
        layout = QVBoxLayout()
        self.setLayout(layout)
        
        self.time_label = QLabel(self)
        self.time_label.setAlignment(Qt.AlignCenter)
        layout.addWidget(self.time_label)

        # 2. Apply styling EXACTLY as seen in your image
        self.time_label.setStyleSheet("font-size: 150px;\n"
                                      "color: hsl(111, 100%, 50%);")
        self.setStyleSheet("background-color: black;")

        # 3. Load the custom digital font from the image
        font_id = QFontDatabase.addApplicationFont("DS-DIGIT.TTF")
        
        # Check if the font loaded successfully (-1 means it failed)
        if font_id != -1:
            font_family = QFontDatabase.applicationFontFamilies(font_id)[0]
            custom_font = QFont(font_family)
            self.time_label.setFont(custom_font)
        else:
            print("Warning: 'DS-DIGIT.TTF' not found in the directory. Falling back to default font.")

        # 4. Timer Logic
        self.timer = QTimer(self)
        self.timer.timeout.connect(self.update_time)
        self.timer.start(1000)

        # Update immediately on startup
        self.update_time()

    def update_time(self):
        # Formats the time to Hours:Minutes:Seconds AM/PM (e.g., 07:57:02 AM)
        current_time = QTime.currentTime().toString("hh:mm:ss AP")
        self.time_label.setText(current_time)


if __name__ == "__main__":
    app = QApplication(sys.argv)
    clock = DigitalClock()
    clock.show()
    sys.exit(app.exec_())
```
