```python
import sys
from PyQt5.QtWidgets import QApplication, QMainWindow, QWidget, QVBoxLayout, QPushButton, QLabel
from PyQt5.QtCore import Qt

class MainWindow(QMainWindow):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("Modern PyQt5 App")
        self.resize(500, 500)
        
        # Track state to toggle between Hello & Goodbye
        self.is_hello = True
        
        self.initUI()

    def initUI(self):
        # 1. Central Widget & Responsive Layout
        central_widget = QWidget(self)
        self.setCentralWidget(central_widget)
        
        layout = QVBoxLayout()
        layout.setAlignment(Qt.AlignCenter)
        layout.setSpacing(30)
        central_widget.setLayout(layout)

        # 2. Modern Styled Label
        self.label = QLabel("Hello", self)
        self.label.setAlignment(Qt.AlignCenter)
        self.label.setStyleSheet("""
            QLabel {
                color: #F8FAFC;
                font-size: 48px;
                font-weight: bold;
                font-family: 'Segoe UI', sans-serif;
            }
        """)

        # 3. Modern Styled Button with Hover & Click Effects
        self.button = QPushButton("Click me!", self)
        self.button.setCursor(Qt.PointingHandCursor)
        self.button.setFixedSize(220, 60)
        self.button.setStyleSheet("""
            QPushButton {
                background-color: #6366F1;
                color: #FFFFFF;
                font-size: 20px;
                font-weight: 600;
                font-family: 'Segoe UI', sans-serif;
                border: none;
                border-radius: 14px;
            }
            QPushButton:hover {
                background-color: #4F46E5;
            }
            QPushButton:pressed {
                background-color: #3730A3;
            }
        """)
        self.button.clicked.connect(self.on_click)

        # 4. Add Widgets to Layout
        layout.addWidget(self.label)
        layout.addWidget(self.button)

        # 5. Application Dark Theme Background
        self.setStyleSheet("""
            QMainWindow {
                background-color: #0F172A;
            }
        """)

    def on_click(self):
        # Toggles seamlessly back and forth
        if self.is_hello:
            self.label.setText("Goodbye")
            self.button.setText("Say Hello!")
        else:
            self.label.setText("Hello")
            self.button.setText("Click me!")
        
        self.is_hello = not self.is_hello


def main():
    app = QApplication(sys.argv)
    window = MainWindow()
    window.show()
    sys.exit(app.exec_())

if __name__ == "__main__":
    main()
```
