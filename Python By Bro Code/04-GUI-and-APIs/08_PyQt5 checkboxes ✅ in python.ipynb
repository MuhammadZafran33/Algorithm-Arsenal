```python
# PyQt5 Checkboxes
import sys
from PyQt5.QtWidgets import QApplication, QMainWindow, QCheckBox
from PyQt5.QtCore import Qt


class MainWindow(QMainWindow):

    def __init__(self):
        super().__init__()
        self.setGeometry(700, 300, 500, 500)
        self.checkbox = QCheckBox("Do you like food?", self)
        self.initUI()

    def initUI(self):
        self.checkbox.setGeometry(10, 0, 500, 100)
        self.checkbox.setStyleSheet("font-size: 30px;" "font-family: Arial;")
        self.checkbox.setChecked(False)
        self.checkbox.stateChanged.connect(self.checkbox_changed)

    def checkbox_changed(self, state):
        if state == Qt.Checked:
            print("You like food")
        else:
            print("You DO NOT like food")


def main():
    app = QApplication(sys.argv)
    window = MainWindow()
    window.show()
    sys.exit(app.exec_())


if __name__ == "__main__":
    main()
```

```python
import sys
from PyQt5.QtWidgets import QApplication, QMainWindow, QWidget, QVBoxLayout, QCheckBox, QLabel
from PyQt5.QtCore import Qt

class MainWindow(QMainWindow):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("Modern PyQt5 Checkbox")
        self.resize(500, 400)
        self.initUI()

    def initUI(self):
        # 1. Central Widget & Centered Layout
        central_widget = QWidget(self)
        self.setCentralWidget(central_widget)

        layout = QVBoxLayout()
        layout.setAlignment(Qt.AlignCenter)
        layout.setSpacing(25)
        central_widget.setLayout(layout)

        # 2. Modern Checkbox
        self.checkbox = QCheckBox("Do you like food?", self)
        self.checkbox.setChecked(False)
        self.checkbox.setCursor(Qt.PointingHandCursor)
        self.checkbox.setStyleSheet("""
            QCheckBox {
                font-size: 22px;
                font-family: 'Segoe UI', sans-serif;
                color: #F8FAFC;
                spacing: 12px;
            }
            QCheckBox::indicator {
                width: 24px;
                height: 24px;
                border-radius: 6px;
                border: 2px solid #6366F1;
                background-color: #1E293B;
            }
            QCheckBox::indicator:hover {
                border-color: #818CF8;
            }
            QCheckBox::indicator:checked {
                background-color: #6366F1;
                border-color: #6366F1;
            }
        """)
        self.checkbox.stateChanged.connect(self.checkbox_changed)

        # 3. Dynamic Status Label
        self.status_label = QLabel("You DO NOT like food ❌", self)
        self.status_label.setAlignment(Qt.AlignCenter)
        self.status_label.setStyleSheet("""
            QLabel {
                font-size: 20px;
                font-family: 'Segoe UI', sans-serif;
                font-weight: 600;
                color: #94A3B8;
            }
        """)

        # 4. Add Widgets
        layout.addWidget(self.checkbox)
        layout.addWidget(self.status_label)

        # 5. Window Background Theme
        self.setStyleSheet("""
            QMainWindow {
                background-color: #0F172A;
            }
        """)

    def checkbox_changed(self, state):
        if state == Qt.Checked:
            self.status_label.setText("You like food! 🍕")
            self.status_label.setStyleSheet("color: #22C55E; font-size: 20px; font-weight: bold;")
            print("You like food")
        else:
            self.status_label.setText("You DO NOT like food ❌")
            self.status_label.setStyleSheet("color: #EF4444; font-size: 20px; font-weight: bold;")
            print("You DO NOT like food")


def main():
    app = QApplication(sys.argv)
    window = MainWindow()
    window.show()
    sys.exit(app.exec_())

if __name__ == "__main__":
    main()
```
