```python
import sys
from PyQt5.QtWidgets import QApplication, QWidget, QLabel, QVBoxLayout, QHBoxLayout, QPushButton
from PyQt5.QtCore import QTimer, Qt
from PyQt5.QtGui import QFont, QFontDatabase

class Stopwatch(QWidget):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("Digital Stopwatch")
        self.resize(600, 250)
        
        # Variables to track time and state
        self.counter = 0
        self.is_running = False
        
        self.initUI()

    def initUI(self):
        # 1. Main Vertical Layout
        main_layout = QVBoxLayout()
        main_layout.setContentsMargins(20, 20, 20, 20)
        self.setLayout(main_layout)
        
        # Apply the black background to the whole window
        self.setStyleSheet("background-color: black;")

        # 2. Stopwatch Display Label
        self.time_label = QLabel("00:00:00.00", self)
        self.time_label.setAlignment(Qt.AlignCenter)
        self.time_label.setStyleSheet("font-size: 120px; color: hsl(111, 100%, 50%);")
        
        # Load the custom digital font
        font_id = QFontDatabase.addApplicationFont("DS-DIGIT.TTF")
        if font_id != -1:
            font_family = QFontDatabase.applicationFontFamilies(font_id)[0]
            self.time_label.setFont(QFont(font_family))
        else:
            print("Warning: 'DS-DIGIT.TTF' not found. Falling back to default font.")

        main_layout.addWidget(self.time_label)

        # 3. Buttons Layout (Horizontal)
        button_layout = QHBoxLayout()
        button_layout.setSpacing(15)
        
        self.start_btn = QPushButton("START", self)
        self.stop_btn = QPushButton("STOP", self)
        self.reset_btn = QPushButton("RESET", self)

        # Style the buttons to look modern and fit the dark theme
        button_style = """
            QPushButton {
                background-color: #1E293B;
                color: #F8FAFC;
                font-size: 24px;
                font-weight: bold;
                font-family: Arial, sans-serif;
                border: 2px solid hsl(111, 100%, 30%);
                border-radius: 10px;
                padding: 10px 0px;
            }
            QPushButton:hover {
                background-color: hsl(111, 100%, 20%);
                border-color: hsl(111, 100%, 50%);
            }
            QPushButton:pressed {
                background-color: hsl(111, 100%, 40%);
            }
        """

        # Apply styles and cursors to buttons
        for btn in (self.start_btn, self.stop_btn, self.reset_btn):
            btn.setStyleSheet(button_style)
            btn.setCursor(Qt.PointingHandCursor)
            button_layout.addWidget(btn)

        main_layout.addLayout(button_layout)

        # 4. Connect Buttons to Functions
        self.start_btn.clicked.connect(self.start_watch)
        self.stop_btn.clicked.connect(self.stop_watch)
        self.reset_btn.clicked.connect(self.reset_watch)

        # 5. Timer Logic (Fires every 10 milliseconds)
        self.timer = QTimer(self)
        self.timer.timeout.connect(self.update_display)

    def start_watch(self):
        # Start timer if it isn't already running
        if not self.is_running:
            self.timer.start(10) 
            self.is_running = True

    def stop_watch(self):
        # Pause the timer
        if self.is_running:
            self.timer.stop()
            self.is_running = False

    def reset_watch(self):
        # Stop timer and reset counter to zero
        self.timer.stop()
        self.is_running = False
        self.counter = 0
        self.time_label.setText("00:00:00.00")

    def update_display(self):
        # Add 1 to counter every 10 milliseconds
        self.counter += 1
        
        # Calculate time components based on the counter
        hours = (self.counter // 360000)
        minutes = (self.counter // 6000) % 60
        seconds = (self.counter // 100) % 60
        hundredths = self.counter % 100
        
        # Format the string to look like HH:MM:SS.ms (e.g., 00:01:23.45)
        time_string = f"{hours:02}:{minutes:02}:{seconds:02}.{hundredths:02}"
        self.time_label.setText(time_string)


if __name__ == "__main__":
    app = QApplication(sys.argv)
    window = Stopwatch()
    window.show()
    sys.exit(app.exec_())
```
