```python
import sys
from PyQt5.QtWidgets import (
    QApplication,
    QMainWindow,
    QPushButton,
    QWidget,
    QHBoxLayout,
)
from PyQt5.QtCore import Qt


class MainWindow(QMainWindow):

    def __init__(self):
        super().__init__()
        self.setWindowTitle("Modern PyQt5 Custom Styled Buttons")
        self.resize(600, 300)

        # Create Buttons
        self.button1 = QPushButton("Red (#1)")
        self.button2 = QPushButton("Green (#2)")
        self.button3 = QPushButton("Blue (#3)")

        self.initUI()

    def initUI(self):
        # 1. Central Widget & Layout
        central_widget = QWidget(self)
        self.setCentralWidget(central_widget)

        hbox = QHBoxLayout()
        hbox.setSpacing(20)
        hbox.setContentsMargins(40, 40, 40, 40)
        central_widget.setLayout(hbox)

        # 2. Add widgets to layout and set hand cursor
        for btn in (self.button1, self.button2, self.button3):
            btn.setCursor(Qt.PointingHandCursor)
            btn.setFixedHeight(80)
            hbox.addWidget(btn)

        # 3. Assign Object Names for QSS ID Selectors (#button1, #button2, #button3)
        self.button1.setObjectName("button1")
        self.button2.setObjectName("button2")
        self.button3.setObjectName("button3")

        # 4. Modern QSS Stylesheet
        self.setStyleSheet("""
            QMainWindow {
                background-color: #0F172A;
            }
            
            /* Base Style for All Buttons */
            QPushButton {
                font-size: 20px;
                font-weight: bold;
                font-family: 'Segoe UI', sans-serif;
                color: #FFFFFF;
                border: none;
                border-radius: 16px;
            }

            /* --- Custom ID Styles --- */
            QPushButton#button1 {
                background-color: hsl(0, 85%, 60%);
            }
            QPushButton#button1:hover {
                background-color: hsl(0, 85%, 72%);
            }
            QPushButton#button1:pressed {
                background-color: hsl(0, 85%, 48%);
            }

            QPushButton#button2 {
                background-color: hsl(145, 65%, 48%);
            }
            QPushButton#button2:hover {
                background-color: hsl(145, 65%, 60%);
            }
            QPushButton#button2:pressed {
                background-color: hsl(145, 65%, 38%);
            }

            QPushButton#button3 {
                background-color: hsl(215, 85%, 60%);
            }
            QPushButton#button3:hover {
                background-color: hsl(215, 85%, 72%);
            }
            QPushButton#button3:pressed {
                background-color: hsl(215, 85%, 48%);
            }
        """)


def main():
    app = QApplication(sys.argv)
    window = MainWindow()
    window.show()
    sys.exit(app.exec_())


if __name__ == "__main__":
    main()
```
