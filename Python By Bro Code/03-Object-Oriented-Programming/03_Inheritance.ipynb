```python
# INHERITANCE =  Allows a class to inherit attribut and methods from another class 
#                Helps with code reusibility and exensibility
#                class child(Parent)
#  
```

```python
class animal:
    def __init__(self, name):
        self.name = name
        self.is_alive = True
    def eat(self):
        print(f"{self.name} is eating..")

    def sleeping(self):
        print(f"{self.name} is sleeping also")

class Dog(animal):
    pass

class Cat(animal):
    pass

class Mouse(animal):
    pass
dog = Dog("scooby")
cat = Cat("Garfield")
mouse = Mouse("Micky")

print("****************")
print(f" name:{dog.name}")
print(f"is alive?: {dog.is_alive}")
dog.eat()
dog.sleeping()


print("****************")
print(f" name:{cat.name}")
print(f"is alive?: {cat.is_alive}")
cat.eat()
cat.sleeping()

print("****************")
print(f" name:{mouse.name}")
print(f"is alive?: {mouse.is_alive}")
mouse.eat()
mouse.sleeping()
```
