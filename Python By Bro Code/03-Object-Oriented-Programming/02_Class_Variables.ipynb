```python
# Class variables  = shared among all the instances of a class 
#       Defined outside the constructor 
#       allow you to share data among all objects created from that class
```

```python
class student:
    class_year = 2026
    def __init__(self, name, age):
        self.name = name
        self.age =  age
student1 = student("Abdul", 18)
student2 = student("zafii", 22)
student3 = student("munib", 24)

print("***********")
print(student1.name)
print(student1.age)
print(student1.class_year)

print("***********")
print(student2.name)
print(student2.age)
print(student2.class_year)

print("***********")
print(student3.name)
print(student3.age)
print(student3.class_year)
```
