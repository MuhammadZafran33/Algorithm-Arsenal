```python
import datetime

# --- Dates ---
date = datetime.date(2025, 1, 2)
today = datetime.date.today()

print(f"Specific date: {date}")
print(f"Today's date: {today}\n")

# --- Times ---
time = datetime.time(12, 30, 0)
now = datetime.datetime.now()

print(f"Specific time: {time}")
print(f"Current date & time (raw): {now}")

# --- Formatting ---
now_formatted = now.strftime("%H:%M:%S %m-%d-%Y")
print(f"Formatted date & time: {now_formatted}\n")

# --- Comparison ---
# target_datetime = datetime.datetime(2020, 1, 2, 12, 30, 1)
target_datetime = datetime.datetime(2027, 1, 2, 12, 30, 1)
current_datetime = datetime.datetime.now()

if target_datetime < current_datetime:
    print("Target date has passed")
else:
    print("Target date has NOT passed")
```
