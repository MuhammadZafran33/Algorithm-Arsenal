```python
# @property = Decorator used to define a method as a property (it can be accessed like an attribute)
#             Benefit: Add additional logic when read, write, or delete attributes
#             Gives you getter, setter, and deleter method

class Rectangle:
    def __init__(self, width, height):
        self._width = width
        self._height = height

    # --- GETTERS ---
    @property
    def width(self):
        return f"{self._width:.1f}cm"

    @property
    def height(self):
        return f"{self._height:.1f}cm"

    # --- SETTERS ---
    @width.setter
    def width(self, new_width):
        if new_width > 0:
            self._width = new_width
        else:
            print("Width must be greater than zero")

    @height.setter
    def height(self, new_height):
        if new_height > 0:
            self._height = new_height
        else:
            print("Height must be greater than zero")

    # --- DELETERS ---
    @width.deleter
    def width(self):
        del self._width
        print("Width has been deleted")

    @height.deleter
    def height(self):
        del self._height
        print("Height has been deleted")


def main():
    # 1. Instantiate the object
    rectangle = Rectangle(3, 4)

    # 2. Test Getters (They act like attributes, no parentheses needed!)
    print("--- Initial Dimensions ---")
    print(rectangle.width)
    print(rectangle.height)

    # 3. Test Setters (Valid values)
    print("\n--- Setting New Dimensions ---")
    rectangle.width = 5
    rectangle.height = 6
    print(rectangle.width)
    print(rectangle.height)

    # 4. Test Setters (Invalid values to trigger error messages)
    print("\n--- Testing Validation Logic ---")
    rectangle.width = -2
    rectangle.height = 0

    # 5. Test Deleters
    print("\n--- Deleting Attributes ---")
    del rectangle.width
    del rectangle.height

if __name__ == '__main__':
    main()
```
