```python
# "Duck typing" = Another way to achieve polymorphism besides Inheritance
#                 Object must have the minimum necessary attributes/methods
#                 "If it looks like a duck and quacks like a duck, it must be a duck."
```

```python
class Animal:
    alive = True

class Dog(Animal):
    def speak(self):
        print("WOOF!")

class Cat(Animal):
    def speak(self):
        print("MEOW!")

class Car:
    alive = False
    
    def speak(self):
        print("HONK!")

# The Car object is added to the list of animals
animals = [Dog(), Cat(), Car()]

# Because of duck typing, the loop works for everything in the list
for animal in animals:
    animal.speak()
    print(animal.alive)
```
