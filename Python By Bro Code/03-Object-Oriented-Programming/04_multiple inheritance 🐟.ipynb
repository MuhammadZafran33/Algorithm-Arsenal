```python
# Multiple Inheritance =  inherit from  more then one parent class 
#                         C(A, B)
# MultiLevel Inheritance = Inherit from a parent which  inherit from another parent 
#                          C(B) <- B(A) <- A
```

```python
class animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        print(f" {self.name} is eating..")
    def sleep(self):
        print(f" {self.name} is sleeping.. ")
# Parent classes defining the inherited behaviors
class Prey(animal):
    def flee(self):
        print(f" {self.name} is fleeing")

class Predator(animal):
    def hunt(self):
        print(f" {self.name} is hunting")

# Child classes inheriting from the parent classes
class Rabbit(Prey):
    pass

class Hawk(Predator):
    pass

# Fish inherits from both (Multiple Inheritance)
class Fish(Prey, Predator):
    pass

# Creating objects (instances) of the classes
rabbit = Rabbit("Bugs")
hawk = Hawk("Tony")
fish = Fish("Nemo")

# Calling the inherited method
rabbit.flee()
rabbit.eat()
rabbit.sleep()
print("-----------------")
hawk.hunt()
hawk.eat()
hawk.sleep()

print("------------------")

fish.flee()
fish.hunt()
fish.eat()
fish.sleep()

```
